```python
iris = sns.load_dataset("iris")
iris
```

```python
iris["species"].value_counts()
```

```python
iris["species"].replace({"setosa": 0, "virginica":1, "versicolor":2})
```

```python
# plt.scatter(iris["sepal_length"], iris["sepal_width"], c = iris["species"], cmap = "winter", marker = "d", label = iris["species"])
# plt.legend()
# legend cannot be made like below this here
# Matplotlib does not automatically know that 0=Setosa, 1=Versicolor, 2=Virginica.
# Instead, it just plots colors for codes.
# If you want a legend with species names, you must add it manually:
```

```python
plt.scatter(iris["sepal_length"], iris["petal_width"], c = iris["species"].astype("category").cat.codes, cmap= "viridis"
```

```python
## adding legend manually

species = np.unique(iris["species"])
color = ['red','green','yellow']

for sp , col in zip(species,color):
    df = iris[iris["species"]==sp] # maked df of each separate species so now scatter plot is made for individual species
    plt.scatter(df["sepal_length"] , df["sepal_width"] , c = col, label = sp)  
    plt.legend()
    # plt.show() this will make separate chart
plt.show() # makes single plot for all df
```

```python
plt.figure(figsize=(10,10))
for i in range(iris.shape[0]):
    plt.text(iris["sepal_length"][i],iris["petal_length"][i], iris["species"][i], fontdict = {"size":8, "color": "red"}, )
plt.axhline(3, color = "green")
plt.axvline(6, color= "green")
plt.xlabel("sepal_lenth")
plt.ylabel("petal_lenth")
plt.scatter( iris["sepal_length"], iris["petal_length"], s = iris["petal_length"]*5, c = iris["species"].astype("category").cat.codes, cmap = "viridis")
```
