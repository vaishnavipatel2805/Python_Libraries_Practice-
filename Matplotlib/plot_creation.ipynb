## ways to create plots:-
1. plt.plot()
2. plt.subplot()
3. plt.subplots()
4. fig.add_subplot()

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. plt.plot()

```python
batter = pd.read_csv("batter.csv")
batter.sample()
```

#### 2d plot

```python

plt.scatter( batter["strike_rate"], batter["avg"], s = batter["runs"])
plt.title("avg vs strike_rate")
plt.xlabel("strike_rate")
plt.ylabel("avg")
plt.show()
```

#### 3d plot


```python

#creating 3d axis first
ax = plt.axes(projection = '3d')
ax.scatter( batter["strike_rate"], batter["avg"] , batter["runs"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg")
```

## 2. plt.subplot()

#### 2d plot:

```python
plt.figure(figsize = (4,4))

ax = plt.subplot() ## only one subplot
ax.scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg")
```

#### 2d subplots:-

```python
## creating subplots but it is reading one subplot at a time

plt.figure(figsize = (10,8))

ax = plt.subplot(2,2,1,)
ax.scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg")

ax = plt.subplot(2,2,2)
ax.scatter(batter["strike_rate"], batter["runs"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("runs")
ax.set_title("strike rate vs runs")
```

```python
plt.figure(figsize = (8,3))

ax = plt.subplot(1,2,1,)
ax.scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg")

ax = plt.subplot(2,2,2)
ax.scatter(batter["strike_rate"], batter["runs"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("runs")
ax.set_title("strike rate vs runs")
```

#### 3d plot:

```python
plt.figure(figsize = (10,10))

ax = plt.subplot(2,2,2 , projection = "3d")
ax.scatter3D(batter["strike_rate"], batter["avg"] , batter["runs"])
ax.set_xlabel("strike rate")
ax.set_zlabel("runs")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg vs avg")
```

#### 2d and 3d subplots mix

```python
plt.figure(figsize = (10,10))

# 1st 2d subplot
ax = plt.subplot(2,2,1)
ax.scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg")

# 2nd 3d subplot
ax = plt.subplot(2,2,2 , projection = "3d")
ax.scatter3D(batter["strike_rate"], batter["avg"] , batter["runs"])
ax.set_xlabel("strike rate")
ax.set_zlabel("runs")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg vs avg")

# 3rd 2d subplot
ax = plt.subplot(2,2,3)
ax.scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg")

# 4th 3d subplot
ax = plt.subplot(2,2,4 , projection = "3d")
ax.scatter3D(batter["strike_rate"], batter["avg"] , batter["runs"])
ax.set_xlabel("strike rate")
ax.set_zlabel("runs")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg vs avg")
```

## 3. plt.subplots()

#### 2d plot

```python
fig,ax = plt.subplots()
ax.scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg")
```

#### 3d plot


```python

fig,ax = plt.subplots(subplot_kw = {"projection" : "3d"})
ax.scatter(batter["strike_rate"], batter["avg"] , batter["runs"], c = batter["runs"])
ax.set_xlabel("strike rate")
ax.set_zlabel("runs")
ax.set_ylabel("avg")
ax.set_title("strike rate vs avg vs avg")
```

#### subplots:- 


```python

fig , ax = plt.subplots(nrows = 2)
ax[0].scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax[1].scatter(batter["strike_rate"], batter["runs"], c = batter["runs"])
```

```python
fig , ax = plt.subplots(ncols = 3)
ax[0].scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax[1].scatter(batter["strike_rate"], batter["runs"], c = batter["runs"])
```

```python
fig , ax = plt.subplots(2,1, sharex = True)
ax[0].scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax[1].scatter(batter["strike_rate"], batter["runs"], c = batter["runs"])
ax[1].set_xlabel("strike rate")
ax[1].set_ylabel("runs")
ax[0].set_ylabel("avg")
```

#### 2d and 3d subplots mix


```python

fig ,ax = plt.subplots(2,2 , subplot_kw = {"projection" : "3d"}, figsize = (10,10))

ax[0,0].scatter3D(batter["strike_rate"], batter["avg"] , batter["runs"], c = batter["runs"])
ax[0,0].set_xlabel("strike rate")
ax[0,0].set_ylabel("avg")
ax[0,0].set_zlabel("runs")

ax[0,1].scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax[0,1].set_xlabel("strike rate")
ax[0,1].set_ylabel("avg")

ax[1,0].scatter3D(batter["strike_rate"], batter["avg"] , batter["runs"], c = batter["runs"])
ax[1,0].set_xlabel("strike rate")
ax[1,0].set_ylabel("avg")
ax[1,0].set_zlabel("runs")

ax[1,1].scatter(batter["strike_rate"], batter["avg"] , c = batter["runs"])
ax[1,1].set_xlabel("strike rate")
ax[1,1].set_ylabel("avg")
```

## 4. fig.add_subplot()

```python
fig = plt.figure()
ax = fig.add_subplot()
ax.scatter(batter["strike_rate"], batter["runs"], c = batter["avg"])
```

```python
ax1 = fig.add_subplot(2,1,1)
ax1.scatter(batter["strike_rate"],batter["runs"])

ax2 = fig.add_subplot(2,1,2)
ax2.scatter(batter["strike_rate"],batter["runs"])

fig.show()
```
