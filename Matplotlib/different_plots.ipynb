```python
iris["species"].replace({"setosa": 0, "virginica":1, "versicolor":2})
```

```python
# plt.scatter(iris["sepal_length"], iris["sepal_width"], c = iris["species"], cmap = "winter", marker = "d", label = iris["species"])
# plt.legend()
# legend cannot be made like below this here
# Matplotlib does not automatically know that 0=Setosa, 1=Versicolor, 2=Virginica.
# Instead, it just plots colors for codes.
# If you want a legend with species names, you must add it manually:
```

```python
plt.scatter(iris["sepal_length"], iris["petal_width"], c = iris["species"].astype("category").cat.codes, cmap= "viridis"
```

```python
## adding legend manually

species = np.unique(iris["species"])
color = ['red','green','yellow']

for sp , col in zip(species,color):
    df = iris[iris["species"]==sp] # maked df of each separate species so now scatter plot is made for individual species
    plt.scatter(df["sepal_length"] , df["sepal_width"] , c = col, label = sp)  
    plt.legend()
    # plt.show() this will make separate chart
plt.show() # makes single plot for all df
```

#### two methods to make any chart:-
1. plt.chart_name()
2. plt.subplots()

```python
movies.sample()
```

```python
plt.style.use("default")
```

```python
# method 1:-

df = movies.groupby("year_of_release")["imdb_votes"].sum().reset_index()
plt.bar(df["year_of_release"], df["imdb_votes"])
plt.title("year of release vs imdb votes")
plt.xlabel("imdb votes")
plt.ylabel("year of releas")
plt.grid()
```

```python
# method 2:-

fig, ax = plt.subplots()
ax.barh(df["year_of_release"], df["imdb_votes"])
ax.set_title("year of release vs imdb votes")
ax.set_ylabel("year of release")
ax.set_xlabel("imdb votes")
ax.grid()
```

```python
df = movies.groupby("year_of_release")["imdb_votes"].mean().reset_index()
temp = movies.groupby("year_of_release")["imdb_rating"].mean().reset_index()
fig , ax = plt.subplots(nrows = 2, figsize = (9,10))
ax[0].barh(df["year_of_release"],df["imdb_votes"], color = "green")
ax[0].set_title("year_of_release vs imdb_votes")
ax[1].bar(temp["year_of_release"],temp["imdb_rating"], color = "pink")
ax[1].set_title("year_of_release vs imdb_ratings")
```

```python
fig , ax = plt.subplots(nrows = 2, ncols = 2, figsize = (9,10))
ax[0,0].barh(df["year_of_release"],df["imdb_votes"], color = "green")
ax[0,0].set_title("year_of_release vs imdb_votes")
ax[0,0].set_xlabel("imdb_votes")
ax[0,0].set_ylabel("year_of_release")

ax[0,1].bar(temp["year_of_release"],temp["imdb_rating"], color = "pink")
ax[0,1].set_title("year_of_release vs imdb_ratings")
ax[0,1].set_ylabel("imdb_ratings")
ax[0,1].set_xlabel("year_of_release")

ax[1,0].barh(df["year_of_release"],df["imdb_votes"], color = "green")
ax[1,0].set_title("year_of_release vs imdb_votes")
ax[1,0].set_xlabel("imdb_votes")
ax[1,0].set_ylabel("year_of_release")


ax[1,1].hist(movies["runtime"])
ax[1,1].set_title("runtime")
ax[1,1].set_ylabel("frequency")
ax[1,1].set_xlabel("runtime")
```
